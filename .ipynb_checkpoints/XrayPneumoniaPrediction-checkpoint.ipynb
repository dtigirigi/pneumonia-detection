# Healthcare Analytics Capstone Project

## Project Objectives
1. Develop a classification model to predict Pneumonia vs Normal from chest X-ray images.
2. Develop a web application/dashboard to accept an X-ray image as input and return the probability of Pneumonia.

## PHASE 1 - Data Preparation and Feature Engineering

```python
# Load libraries
import os
import cv2
import numpy as np
import pandas as pd
from PIL import Image
import mahotas as mh

# Folder paths
BASE = os.path.join("Dataset", "Chest Xray Images Full")

TRAIN_DIR = os.path.join(BASE, "Train")
VAL_DIR   = os.path.join(BASE, "Val")
TEST_DIR  = os.path.join(BASE, "Test")

IMG_SIZE = (224, 224)

# Load Metadata
def load_metadata(root):
    rows = []
    for label in ["NORMAL", "PNEUMONIA"]:
        folder = os.path.join(root, label)
        for f in os.listdir(folder):
            if f.lower().endswith((".jpg", ".jpeg", ".png")):
                rows.append([f, label, os.path.join(folder, f)])
    return pd.DataFrame(rows, columns=["Image_ID", "Label", "File_Path"])

train_meta = load_metadata(TRAIN_DIR)
val_meta   = load_metadata(VAL_DIR)
test_meta  = load_metadata(TEST_DIR)

# Check for corrupted images
def find_bad_images(df):
    bad = []
    for path in df["File_Path"]:
        try:
            Image.open(path)
        except Exception as e:
            bad.append((path, str(e)))
    return bad

bad_train = find_bad_images(train_meta)
bad_val   = find_bad_images(val_meta)
bad_test  = find_bad_images(test_meta)

print("Bad TRAIN images:", bad_train)
print("Bad VAL images:", bad_val)
print("Bad TEST images:", bad_test)

# Remove corrupted images
def remove_bad(df, bad_list):
    bad_paths = [b[0] for b in bad_list]
    return df[~df["File_Path"].isin(bad_paths)]

train_meta = remove_bad(train_meta, bad_train)
val_meta   = remove_bad(val_meta, bad_val)
test_meta  = remove_bad(test_meta, bad_test)

# Feature Extraction Functions
def extract_intensity_features(img):
    return {
        "Mean_Intensity": np.mean(img),
        "Std_Intensity": np.std(img),
        "Min_Intensity": np.min(img),
        "Max_Intensity": np.max(img)
    }

def extract_haralick_features(img):
    img_uint8 = img.astype(np.uint8) # grayscale uint8 for Mohatas
    har = mh.features.haralick(img_uint8)
    har_mean = har.mean(axis=0)
    return {f"Haralick_{i+1}": har_mean[i] for i in range(len(har_mean))}

# Extract Features for ALL images (Train + Val + Test)
def extract_features(df):
    feature_rows = []

    for _, row in df.iterrows():
        path = row["File_Path"]
        label = row["Label"]

        img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
        img = cv2.resize(img, IMG_SIZE)

        intensity = extract_intensity_features(img)
        haralick = extract_haralick_features(img)

        all_features = {
            "Image_ID": row["Image_ID"],
            "Label": label,
            **intensity,
            **haralick
        }

        feature_rows.append(all_features)

    return pd.DataFrame(feature_rows)

train_features = extract_features(train_meta)
val_features   = extract_features(val_meta)
test_features  = extract_features(test_meta)

# Build Master Dataset for Train 
master_df = train_features.copy()

# Binary encode label 
master_df["Label_Binary"] = (master_df["Label"] == "PNEUMONIA").astype(int)

print("\nMaster DF Ready:")
print(master_df.head())
print("\nShape:", master_df.shape)
```

## PHASE 2 - Exploratory Data Analysis (EDA)

```python
# Load libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")

# Count Normal vs Pneumonia and Check class imbalance
plt.figure(figsize=(6,4))
sns.countplot(data=master_df, x="Label", palette="viridis")
plt.title("Distribution of Normal vs Pneumonia Cases")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()

print(master_df["Label"].value_counts())
print("\nClass Imbalance (%):")
print(master_df["Label"].value_counts(normalize=True) * 100)

# Feature Analysis
# Summary Statistics Table (Normal vs Pneumonia vs Overall)
num_features = ["Mean_Intensity","Std_Intensity","Min_Intensity","Max_Intensity"]

summary_table = pd.DataFrame()

for feat in num_features:
    summary_table.loc[feat, "Normal_Mean_SD"] = (
        f"{master_df[master_df.Label=='NORMAL'][feat].mean():.2f} "
        f"({master_df[master_df.Label=='NORMAL'][feat].std():.2f})"
    )
    summary_table.loc[feat, "Pneumonia_Mean_SD"] = (
        f"{master_df[master_df.Label=='PNEUMONIA'][feat].mean():.2f} "
        f"({master_df[master_df.Label=='PNEUMONIA'][feat].std():.2f})"
    )
    summary_table.loc[feat, "Overall_Mean_SD"] = (
        f"{master_df[feat].mean():.2f} ({master_df[feat].std():.2f})"
    )

summary_table

# Boxplots by Class
plt.figure(figsize=(12,8))
for i, feat in enumerate(num_features):
    plt.subplot(2,2,i+1)
    sns.boxplot(data=master_df, x="Label", y=feat, palette="coolwarm")
    plt.title(f"{feat} by Class")
plt.tight_layout()
plt.show()

# Create Intensity Categories
master_df["Intensity_Category"] = pd.qcut(
    master_df["Mean_Intensity"],
    q=3,
    labels=["Low Intensity","Medium Intensity","High Intensity"]
)

# Table 2 — Pneumonia Rate by Intensity Category
table2 = master_df.groupby("Intensity_Category").agg(
    N=("Label","count"),
    Pneumonia_Cases=("Label_Binary","sum")
)

table2["Pneumonia_Rate_%"] = (table2["Pneumonia_Cases"] / table2["N"]) * 100
table2


# Bar Chart for Categorical Features
plt.figure(figsize=(6,4))
sns.barplot(data=table2.reset_index(),
            x="Intensity_Category",
            y="Pneumonia_Rate_%",
            palette="magma")
plt.title("Pneumonia Rate by Intensity Category")
plt.ylabel("Pneumonia Rate (%)")
plt.show()

# Multivariable Analysis - interaction of multiple features
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=master_df,
    x="Mean_Intensity",
    y="Haralick_1",
    hue="Label",
    palette="coolwarm",
    alpha=0.7
)
plt.title("Mean Intensity vs Haralick_1 (Texture) by Class")
plt.show()

# Correlation Analysis - Identify redundant features
plt.figure(figsize=(14,10))
corr = master_df[num_features + [f"Haralick_{i+1}" for i in range(13)]].corr()
sns.heatmap(corr, cmap="coolwarm", annot=False)
plt.title("Correlation Matrix of Derived Features")
plt.show()
```

## PHASE 3 - Logistic Regression (With VIF, ROC and Threshold)

```python
# Load libraries
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    roc_curve, roc_auc_score,
    confusion_matrix, accuracy_score
)

import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Data Preparation (using Phase 2 master data)
# Binary encode label: 1 = PNEUMONIA, 0 = NORMAL
master_df["Label_Binary"] = (master_df["Label"] == "PNEUMONIA").astype(int)

# Use all derived features from Phase 2
feature_cols_all = [
    "Mean_Intensity", "Std_Intensity", "Min_Intensity", "Max_Intensity",
]

X_all = master_df[feature_cols_all]
y = master_df["Label_Binary"]

# Train–test split (for logistic regression)
X_train_all, X_test_all, y_train, y_test = train_test_split(
    X_all, y, test_size=0.3, random_state=42, stratify=y
)

# Helper functions

def compute_vif(df):
    vif = pd.DataFrame()
    vif["feature"] = df.columns
    vif["VIF"] = [variance_inflation_factor(df.values, i) for i in range(df.shape[1])]
    return vif

def fit_logit(X, y):
    X_const = sm.add_constant(X)
    model = sm.Logit(y, X_const).fit(disp=False)
    return model

def evaluate_on_set(model, X, y, threshold=0.5, label="TRAIN"):
    X_const = sm.add_constant(X)
    y_prob = model.predict(X_const)
    auc = roc_auc_score(y, y_prob)

    # default or custom threshold
    y_pred = (y_prob >= threshold).astype(int)

    cm = confusion_matrix(y, y_pred)
    tn, fp, fn, tp = cm.ravel()

    acc = accuracy_score(y, y_pred)
    sensitivity = tp / (tp + fn) if (tp + fn) > 0 else np.nan  # recall for PNEUMONIA
    specificity = tn / (tn + fp) if (tn + fp) > 0 else np.nan  # recall for NORMAL

    print(f"\n=== {label} SET EVALUATION (threshold={threshold:.3f}) ===")
    print("AUC:", auc)
    print("Accuracy:", acc)
    print("Sensitivity (PNEUMONIA):", sensitivity)
    print("Specificity (NORMAL):", specificity)
    print("Confusion Matrix:\n", cm)

    return {
        "auc": auc,
        "acc": acc,
        "sens": sensitivity,
        "spec": specificity,
        "threshold": threshold,
        "y_prob": y_prob,
        "y_pred": y_pred
    }

def find_best_threshold(y_true, y_prob):
    fpr, tpr, thresholds = roc_curve(y_true, y_prob)
    youden = tpr - fpr
    idx = np.argmax(youden)
    best_thr = thresholds[idx]
    return best_thr, fpr, tpr, thresholds

# VERSION DEFINITIONS 
versions = {
    "ALL_FEATURES": feature_cols_all,
    "INTENSITY_ONLY": [
        "Mean_Intensity", "Std_Intensity", "Min_Intensity", "Max_Intensity"
    ],
}

results_summary = []

for version_name, feature_cols in versions.items():
    print("\n" + "="*70)
    print(f"PHASE 3 – LOGISTIC REGRESSION VERSION: {version_name}")
    print("="*70)

    # Using training data to run binary logistic regression
    X_train = X_train_all[feature_cols]
    X_test = X_test_all[feature_cols]

    print("\nMulticollinearity check (VIF) on TRAIN features:")
    vif_df = compute_vif(X_train)
    print(vif_df)

    # Check significance (statsmodels summary)
    logit_model = fit_logit(X_train, y_train)
    print("\nLogistic Regression Summary (TRAIN):")
    print(logit_model.summary())

    # ROC/AUC, classification table, accuracy (TRAIN, threshold=0.5)
    train_eval_default = evaluate_on_set(
        logit_model, X_train, y_train, threshold=0.5, label=f"{version_name} – TRAIN (default thr)"
    )

    # Best threshold to balance sensitivity and specificity (TRAIN)
    best_thr, fpr_train, tpr_train, thr_train = find_best_threshold(
        y_train, train_eval_default["y_prob"]
    )
    print(f"\nBest threshold for TRAIN: {best_thr:.4f}")

    train_eval_best = evaluate_on_set(
        logit_model, X_train, y_train, threshold=best_thr, label=f"{version_name} – TRAIN (best thr)"
    )

    # ROC/AUC, accuracy, sensitivity, specificity on TEST using best threshold
    test_eval_best = evaluate_on_set(
        logit_model, X_test, y_test, threshold=best_thr, label=f"{version_name} – TEST (best thr)"
    )

    # Store summary for final model selection
    results_summary.append({
        "version": version_name,
        "train_auc": train_eval_best["auc"],
        "test_auc": test_eval_best["auc"],
        "best_threshold": best_thr,
        "train_acc": train_eval_best["acc"],
        "test_acc": test_eval_best["acc"],
        "train_sens": train_eval_best["sens"],
        "test_sens": test_eval_best["sens"],
        "train_spec": train_eval_best["spec"],
        "test_spec": test_eval_best["spec"],
    })

# Compare test AUC
results_df = pd.DataFrame(results_summary)
print("\n\n=== PHASE 3 – MODEL VERSION COMPARISON (by TEST AUC) ===")
print(results_df.sort_values("test_auc", ascending=False))

best_model_row = results_df.sort_values("test_auc", ascending=False).iloc[0]
print("\nFinal chosen version based on highest TEST AUC:")
print(best_model_row)
```

## PHASE 4 - ML Models Comparison with Logistic Regression

```python
# Load libraries
import os
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier

from sklearn.metrics import (
    roc_curve, roc_auc_score,
    confusion_matrix, accuracy_score
)

import joblib
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")

os.makedirs("models", exist_ok=True)


def eval_and_save(model, X_train, y_train, X_test, y_test, name):
    # Fit model
    model.fit(X_train, y_train)

    # Save model to models
    save_path = os.path.join("models", f"{name}.joblib")
    joblib.dump(model, save_path)
    print(f"Saved: {save_path}")

    # Predict probabilities
    if hasattr(model, "predict_proba"):
        y_train_prob = model.predict_proba(X_train)[:,1]
        y_test_prob  = model.predict_proba(X_test)[:,1]
    else:
        # SVM decision_function fallback
        y_train_prob = model.decision_function(X_train)
        y_test_prob  = model.decision_function(X_test)

        # Normalize to 0–1
        y_train_prob = (y_train_prob - y_train_prob.min()) / (y_train_prob.max() - y_train_prob.min())
        y_test_prob  = (y_test_prob  - y_test_prob.min())  / (y_test_prob.max()  - y_test_prob.min())

    # Train metrics
    train_auc = roc_auc_score(y_train, y_train_prob)
    y_train_pred = (y_train_prob >= 0.5).astype(int)
    cm_train = confusion_matrix(y_train, y_train_pred)

    # Test metrics
    test_auc = roc_auc_score(y_test, y_test_prob)
    y_test_pred = (y_test_prob >= 0.5).astype(int)
    cm_test = confusion_matrix(y_test, y_test_pred)

    print(f"\n=== {name} – TRAIN ===")
    print("AUC:", train_auc)
    print("Confusion Matrix:\n", cm_train)

    print(f"\n=== {name} – TEST ===")
    print("AUC:", test_auc)
    print("Confusion Matrix:\n", cm_test)

    return {
        "model": name,
        "train_auc": train_auc,
        "test_auc": test_auc
    }

# RUN All classical MLs and save them
results_phase4 = []

X_train = X_train_all[feature_cols_all]
X_test  = X_test_all[feature_cols_all]

# Naive Bayes
results_phase4.append(
    eval_and_save(GaussianNB(), X_train, y_train, X_test, y_test, "naive_bayes")
)

# Decision Tree
results_phase4.append(
    eval_and_save(
        DecisionTreeClassifier(random_state=42),
        X_train, y_train, X_test, y_test,
        "decision_tree"
    )
)

# Random Forest
results_phase4.append(
    eval_and_save(
        RandomForestClassifier(n_estimators=200, random_state=42),
        X_train, y_train, X_test, y_test,
        "random_forest"
    )
)

# SVM (with probability=True)
results_phase4.append(
    eval_and_save(
        SVC(kernel="rbf", probability=True, random_state=42),
        X_train, y_train, X_test, y_test,
        "svm_rbf"
    )
)

# XGBoost
results_phase4.append(
    eval_and_save(
        XGBClassifier(
            n_estimators=300,
            max_depth=4,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            random_state=42,
            eval_metric="logloss"
        ),
        X_train, y_train, X_test, y_test,
        "xgboost"
    )
)

# Compare Test AUC and Finalize Best ML Model
phase4_df = pd.DataFrame(results_phase4)
print("\n=== PHASE 4 – TEST AUC COMPARISON ===")
print(phase4_df.sort_values("test_auc", ascending=False))

best_phase4 = phase4_df.sort_values("test_auc", ascending=False).iloc[0]
print("\nBest classical ML method (by TEST AUC):")
print(best_phase4)
```

## PHASE 5 - Deep Learning using CNN

```python
# Load libraries
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint

import absl.logging
absl.logging.set_verbosity(absl.logging.ERROR)

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=10,
    width_shift_range=0.05,
    height_shift_range=0.05,
    zoom_range=0.1,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary"
)

val_gen = val_datagen.flow_from_directory(
    VAL_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary"
)

test_gen = test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    shuffle=False
)

# Build CNN model and perform basic hyperparameter tuning 
def build_cnn_model(
    input_shape=(224,224,3),
    conv_blocks=3,
    base_filters=32,
    dropout_rate=0.5
):
    model = models.Sequential()
    filters = base_filters

    for i in range(conv_blocks):
        model.add(layers.Conv2D(filters, (3,3), activation="relu", padding="same", input_shape=input_shape if i==0 else None))
        model.add(layers.Conv2D(filters, (3,3), activation="relu", padding="same"))
        model.add(layers.MaxPooling2D((2,2)))
        filters *= 2

    model.add(layers.Flatten())
    model.add(layers.Dense(256, activation="relu"))
    model.add(layers.Dropout(dropout_rate))
    model.add(layers.Dense(1, activation="sigmoid"))

    model.compile(
        optimizer=optimizers.Adam(learning_rate=1e-4),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )
    return model

cnn_model = build_cnn_model()

# Train CNN and Save the Model
es = EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
rlr = ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3)

history = cnn_model.fit(
    train_gen,
    epochs=15,
    validation_data=val_gen,
    callbacks=[es, rlr]
)

cnn_model.save("models/cnn_pneumonia_model.h5")

# Evaluate CNN on Test Data
from sklearn.metrics import (
    roc_auc_score, roc_curve,
    confusion_matrix, accuracy_score,
    precision_score, recall_score, f1_score
)
import numpy as np

test_gen.reset()
y_prob_cnn = cnn_model.predict(test_gen).ravel()
y_true_cnn = test_gen.classes

auc_cnn = roc_auc_score(y_true_cnn, y_prob_cnn)
fpr_cnn, tpr_cnn, thr_cnn = roc_curve(y_true_cnn, y_prob_cnn)

best_thr_cnn = thr_cnn[np.argmax(tpr_cnn - fpr_cnn)]
y_pred_cnn = (y_prob_cnn >= best_thr_cnn).astype(int)

cm_cnn = confusion_matrix(y_true_cnn, y_pred_cnn)
tn, fp, fn, tp = cm_cnn.ravel()

acc_cnn = accuracy_score(y_true_cnn, y_pred_cnn)
sens_cnn = recall_score(y_true_cnn, y_pred_cnn)          
spec_cnn = tn / (tn + fp)
prec_cnn = precision_score(y_true_cnn, y_pred_cnn)
f1_cnn = f1_score(y_true_cnn, y_pred_cnn)

print("\n=== CNN – TEST EVALUATION ===")
print("AUC:", auc_cnn)
print("Accuracy:", acc_cnn)
print("Sensitivity:", sens_cnn)
print("Specificity:", spec_cnn)
print("Precision:", prec_cnn)
print("F1-score:", f1_cnn)
print("Confusion Matrix:\n", cm_cnn)
print("Best threshold (Youden):", best_thr_cnn)

# Compare CNN with classical ML 
comparison_rows = []

# Add best logistic regression version from Phase 3
comparison_rows.append({
    "Model": "Logistic Regression (best version)",
    "Test_AUC": best_model_row["test_auc"]
})

# Add Phase 4 models
for row in results_phase4:
    comparison_rows.append({
        "Model": row["model"],
        "Test_AUC": row["test_auc"]
    })

# Add CNN
comparison_rows.append({
    "Model": "CNN",
    "Test_AUC": auc_cnn
})

comparison_df = pd.DataFrame(comparison_rows)
print("\n=== MODEL COMPARISON (TEST AUC) ===")
print(comparison_df.sort_values("Test_AUC", ascending=False))

best_overall = comparison_df.sort_values("Test_AUC", ascending=False).iloc[0]
print("\nBest overall model for deployment:")
print(best_overall)
```
